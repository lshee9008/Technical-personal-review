### Q2. [DB] 트랜잭션과 ACID를 환불 예시로 설명하면?
**내 답:** (여기를 더블클릭해서 적기)

<details><summary>모범 답안</summary>

환불 = 주문 상태 변경 + 포인트 복구 + 환불 기록을 한 트랜잭션으로 묶어요. 하나라도 실패하면 전부 롤백(원자성), 포인트 음수 같은 규칙 위반 상태로 안 끝남(일관성), 동시에 도는 트랜잭션이 중간 상태를 못 봄(격리성), 커밋하면 서버가 꺼져도 남음(지속성).
</details>

### Q3. [DB] MySQL과 MariaDB의 관계, RDB와 MongoDB는 언제 각각 쓰나?
**내 답:** (여기를 더블클릭해서 적기)

<details><summary>모범 답안</summary>

MariaDB는 MySQL 원개발자들이 갈라져 나와 만든 오픈소스로 거의 호환돼요. 정해진 스키마와 트랜잭션으로 정확성이 중요한 주문·결제는 RDB, 구조가 자주 바뀌는 대화 로그·상품 속성은 MongoDB 같은 문서형이 맞아요. 고객지원 에이전트라면 주문은 RDB, 대화 기록은 문서형으로 나눌 수 있어요.
</details>

### Q4. [백엔드] 401과 403의 차이, 세션과 JWT의 차이는?
**내 답:** (여기를 더블클릭해서 적기)

<details><summary>모범 답안</summary>

401은 누구인지 모를 때(인증 필요), 403은 누구인지는 알지만 권한이 없을 때(인가 실패). 세션은 서버가 로그인 상태를 저장하고 브라우저엔 ID 쿠키만 둬서 강제 만료가 쉬워요. JWT는 서명된 토큰을 클라이언트가 들고 다녀 서버 저장이 필요 없지만 만료 전 무효화가 어려워요.
</details>

### Q5. [백엔드] Django ORM의 N+1 문제와 해결법은?

</details>

### Q10. [딥러닝] PyTorch 학습 루프 5단계를 순서대로 쓰고, zero_grad를 빼면 어떻게 되나?
**내 답:** (여기를 더블클릭해서 적기)

<details><summary>모범 답안</summary>

logits=model(X) → loss=loss_fn(logits, y) → opt.zero_grad() → loss.backward() → opt.step(). zero_grad를 빼면 기울기가 계속 누적돼 엉뚱한 방향·크기로 갱신돼요.
</details>

### Q12. [딥러닝] Transformer가 RNN보다 큰 모델 학습에 유리한 이유는?
**내 답:** (여기를 더블클릭해서 적기)

<details><summary>모범 답안</summary>

RNN은 앞 단계가 끝나야 다음을 계산해 병렬화가 안 되고 긴 문장에서 앞을 잊어요. Transformer는 Self-Attention으로 모든 단어 쌍의 관련도를 행렬곱 한 번에 계산해 GPU 병렬 처리가 되고 먼 단어도 바로 연결돼요.
</details>
